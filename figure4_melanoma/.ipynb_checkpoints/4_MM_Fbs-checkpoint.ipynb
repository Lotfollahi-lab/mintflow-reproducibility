```python
if not DATA_PREPARED:
    sc.pp.neighbors(adata_spatial, use_rep = 'X_pca', 
                    metric = "euclidean", n_neighbors=20)#key_added=neighbor_id)
    sc.tl.umap(adata_spatial, min_dist=0.1) #neighbor_key=)

    RES=0.4
    sc.tl.leiden(adata_spatial, #restrict_to=(CATEGORY,subset_cluster), 
                 resolution=RES, random_state=1, key_added=f'leiden_res{RES}') #neighbors_key='neighbor_30')
    
    try:
        RENAME = {'4': '#ffef5a',  
                  '1': '#8981a7',
                  '0': '#91bae2', #"#75fbfd",  
                  '3': '#c6508f',
                  '2': "#2f6565"}

        # make sure the cluster column is categorical
        clust_key = f"leiden_res{RES}"
        adata_spatial.obs[clust_key] = adata_spatial.obs[clust_key].astype("category")
            # build the colour list in the exact order Scanpy expects
        cats   = adata_spatial.obs[clust_key].cat.categories
        pal    = [RENAME.get(c, "#D3D3D3") for c in cats]   # fall back to grey if missing
        adata_spatial.uns[f"{clust_key}_colors"] = pal
    except:
        1



    sc.pl.umap(adata_spatial,  
               color=[ key_celltype, f"leiden_res{RES}"],
               s=200,legend_loc="on data",
               legend_fontsize=30,
               legend_fontoutline=2,
              )

    
    try:
        del(adata_spatial.uns[f"leiden_res{RES}_colors"])
    except:
        1
    {x:x for x in adata_spatial.obs[f"leiden_res{RES}"].unique()}
    RENAME = {'4': 'Superficial-enriched', 
          '1': 'Fibroblast_spatial1',
          '0': 'Fibroblast_spatial2',
          '3': 'Fibroblast_spatial3',
          '2': 'Fibroblast_spatial4',
         }
    adata_spatial.obs["spatial"] =  adata_spatial.obs[f"leiden_res{RES}"].map(RENAME)
    try:
        RENAME = {'Superficial-enriched' : '#ffef5a',  
                  'Fibroblast_spatial1' : '#8981a7',
                  'Fibroblast_spatial2' : '#91bae2', #"#75fbfd",  
                   'Fibroblast_spatial3'  : '#c6508f',
                  'Fibroblast_spatial4' : "#2f6565"}

        # make sure the cluster column is categorical
        clust_key = f"spatial"
        adata_spatial.obs[clust_key] = adata_spatial.obs[clust_key].astype("category")
            # build the colour list in the exact order Scanpy expects
        cats   = adata_spatial.obs[clust_key].cat.categories
        pal    = [RENAME.get(c, "#D3D3D3") for c in cats]   # fall back to grey if missing
        adata_spatial.uns[f"{clust_key}_colors"] = pal
    except:
        1
```
