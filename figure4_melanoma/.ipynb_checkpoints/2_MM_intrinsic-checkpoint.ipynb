```python
# Specify the number of top contributors to display
top_n = 9
GROUP =  f"leiden_res{RES}"

# Group by `niche_name` and `Annotation`, and count occurrences
grouped_data = (
    adata.obs.groupby([GROUP, key_celltype])
    .size()
    .reset_index(name="count")
)

# Calculate the total count for each `niche_name`
grouped_data["total_count"] = grouped_data.groupby(GROUP)["count"].transform("sum")

# Calculate the percentage contribution for each `Annotation` within its `niche_name`
grouped_data["percentage"] = round((grouped_data["count"] / grouped_data["total_count"]) * 100)

# Sort the counts in descending order within each `niche_name` group
sorted_data = grouped_data.sort_values(
    [GROUP, "count"], ascending=[True, False]
)

# Extract the top N contributors for each group
top_contributors = sorted_data.groupby(GROUP).head(top_n)

import pandas as pd

# Remove the display limit
pd.set_option('display.max_rows', None)

# Display the result
top_contributors
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Count cells per (niche_name, lvl5_annotation)
df = ADATA.obs.groupby(["intrinsic", key_celltype]).size().unstack(fill_value=0)

# Normalize to get proportions
df_norm = df.div(df.sum(axis=1), axis=0)

# Filter out low-proportion categories (<0.02)
df_norm = df_norm.loc[:, df_norm.mean() >= 0.002]  # Keep only those with mean proportion >= 0.02

# Get colors for only the categories in df_norm
color_list = [colors.get(cat, "#D3D3D3") for cat in df_norm.columns]  # Default to grey if missing

# Plot stacked bar chart with custom colors
fig, ax = plt.subplots(figsize=(9, 5))
df_norm.plot(kind="bar", stacked=True, color=color_list,
             edgecolor="black", ax=ax, linewidth=0.3)

# Remove grid
ax.grid(False)  

# Remove top and right border
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Remove title
ax.set_title("")

# Labeling
plt.xlabel(" ")
plt.ylabel("Proportion")

# Reverse the legend order
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles[::-1], labels[::-1], title="", bbox_to_anchor=(1.05, 1), loc="upper left", frameon=False)

plt.xticks(rotation=90)

# Show plot
plt.tight_layout()
plt.show()
```
